```python
before=id(Y)
Y=X+Y
id(Y)==before
```

```python
#使用切片表示法将操作的结果分配给先前分配的数组，例如Y[:] = <expression>。
```

```python
Z = torch.zeros_like(Y)
print("idZ:",id(Z))
Z[:] = X + Y
print("idZ:",id(Z))
```

```python
#或者使用X += Y来减少操作的内存开销。
```
